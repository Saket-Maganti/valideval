# ValidEval V5 — Controlled common-panel BBH

Canonical T4×2 notebook. `fixture` mode is deterministic and always labeled `NON_EVIDENCE_FIXTURE`. Real modes require a frozen V5 configuration and do not become evidence until merge, checksum, importer, coverage, and cross-benchmark gates pass. Select Kaggle T4×2 before real execution; secrets must come from Kaggle Secrets or environment variables.


```python
from __future__ import annotations

import os
import subprocess
import sys
from pathlib import Path

from valideval.execution import EXECUTION_MODES, run_notebook_stage

STAGE = "bbh"
MODE = os.environ.get("VALIDEVAL_EXECUTION_MODE", "fixture").strip().lower()
OUTPUT_ROOT = Path(os.environ.get("VALIDEVAL_NOTEBOOK_OUTPUT_ROOT", "kaggle_max_ceiling_outputs"))
if MODE not in EXECUTION_MODES:
    raise ValueError(f"Unsupported execution mode: {MODE}")
if os.environ.get("VALIDEVAL_INSTALL_EDITABLE") == "1":
    subprocess.run([sys.executable, "-m", "pip", "install", "-e", "."], check=True)
```

```python
RESULT = run_notebook_stage(STAGE, mode=MODE, output_root=OUTPUT_ROOT)
assert RESULT["mode"] == MODE
if MODE == "fixture":
    assert RESULT["evidence_state"] == "NON_EVIDENCE_FIXTURE"
print(RESULT["status"])
```
